# Introduction

This project focuses on analyzing ecommerce consumer behavior using a dataset containing detailed information about customer demographics, purchasing habits, and preferences. The goal is to uncover insights into consumer behavior patterns, preferences, and trends that can help businesses make data-driven decisions to improve customer satisfaction, optimize marketing strategies, and enhance overall sales performance.

The dataset includes various attributes such as customer demographics (age, gender, income level, etc.), purchase details (categories, amounts, frequency), and behavioral factors (time spent on research, brand loyalty, satisfaction, etc.). Through exploratory data analysis (EDA), we aim to answer key business questions, such as:

1. Identifying the most popular purchase categories.
2. Understanding the distribution of time spent on product research.
3. Analyzing discount usage and its impact on customer behavior.
4. Exploring the relationship between purchase amount and customer satisfaction.
5. Examining customer preferences for payment methods and shopping devices.

By leveraging visualization techniques and statistical analysis, this project provides actionable insights into ecommerce consumer behavior, enabling businesses to better understand their customers and tailor their strategies accordingly.

Clean Purchase_Amount so we can use it numerically

```python
df['Purchase_Amount'] = df['Purchase_Amount'].replace('[\$,]', '', regex=True).astype(float)
```

Question and Answers


1. What are the top purchase categories?

```python
top_categories = df['Purchase_Category'].value_counts().head(10)
print(top_categories)

top_categories.plot(kind='bar', color='skyblue')
plt.title('Top 10 Purchase Categories')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```


2. What is the distribution of time spent on product research

```python
sns.histplot(df['Time_Spent_on_Product_Research(hours)'], bins=20, kde=True)
plt.title('Time Spent on Product Research')
plt.xlabel('Hours')
plt.ylabel('Frequency')
plt.show()
```


3. How many customers used a discount?

```python
discount_counts = df['Discount_Used'].value_counts()
print(discount_counts)

discount_counts.plot(kind='bar', color='skyblue')
plt.title('Discount Usage')
plt.xticks([0, 1], ['False', 'True'], rotation=0)
plt.ylabel('Customers')
plt.show()
```

4. What is the distribution of product ratings?


```python
sns.countplot(x='Product_Rating', data=df, palette='viridis')
plt.title('Product Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

5. Which payment methods do customers prefer?

```python
payment_methods = df['Payment_Method'].value_counts()
print(payment_methods)

payment_methods.plot(kind='bar', color='skyblue')
plt.title('Payment Methods Used')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

6. What is the relationship between purchase amount and customer satisfaction?

```python
sns.scatterplot(x='Purchase_Amount', y='Customer_Satisfaction', data=df)
plt.title('Purchase Amount vs Customer Satisfaction')
plt.xlabel('Purchase Amount')
plt.ylabel('Satisfaction')
plt.show()
```

7. What is the average purchase frequency by income level?

```python
income_freq = df.groupby('Income_Level')['Frequency_of_Purchase'].mean()
print(income_freq)

income_freq.plot(kind='bar', color='lightblue')
plt.title('Average Purchase Frequency by Income Level')
plt.ylabel('Frequency')
plt.xticks(rotation=45)
plt.show()
```

8. What devices do customers use for shopping?

```python
device_counts = df['Device_Used_for_Shopping'].value_counts()
print(device_counts)

device_counts.plot(kind='bar', color='skyblue')
plt.title('Devices Used for Shopping')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```


9: Which purchase category has the highest average purchase amount?

```python
top_category = df.groupby('Purchase_Category')['Purchase_Amount'].mean().sort_values(ascending=False)
print(top_category.head(1))
```

10: What's the most commonly used device for shopping?

```python
most_common_device = df['Device_Used_for_Shopping'].mode()[0]
print("Most common device:", most_common_device)
```
